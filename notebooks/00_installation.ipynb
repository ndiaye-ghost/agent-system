## Installer un modèle local avec Ollama

1. Télécharger Ollama : https://ollama.com/download
2. Dans un terminal : `ollama pull qwen2.5:14b` (≈ 9 Go, GPU de 10 Go conseillé ; `qwen2.5:7b` pour un PC portable)
3. Ollama démarre automatiquement sur `http://localhost:11434`

Sur un serveur d'entreprise, remplacez `BASE_URL` par l'adresse du serveur.

````python
# Configuration du modèle hébergé (Ollama par défaut)
BASE_URL = "http://localhost:11434/v1"   # vLLM : "http://serveur:8000/v1"
MODEL = "qwen2.5:14b"

from safety_agent.notebook import get_llm
llm = get_llm(BASE_URL, MODEL)   # bascule en MODE DÉMO si le serveur ne répond pas
````

````python
# Test rapide du modèle
rep = llm.chat([{"role": "user", "content": "Réponds en une phrase : qu'est-ce qu'un ASIL ?"}])
print(rep.get("content") or rep)
````
